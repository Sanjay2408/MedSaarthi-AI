# Program 12: LangChain and the Cohere key

**Aim.** Install langchain, cohere and langchain-community, obtain the API key, and run a first chain.

**Theory.** LangChain composes steps with the `|` operator: prompt | model | output parser. The model object is the only part that changes between providers.

```python
import sys
sys.path.insert(0, '.')
from common import *
```

```python
import importlib.metadata as md
for p in ['langchain', 'langchain-cohere', 'langchain-community', 'langchain-groq']:
    print(f'{p:22s}', md.version(p))
print('install command: pip install langchain cohere langchain-cohere langchain-community')
```

```python
import os
print('COHERE_API_KEY set:', bool(os.getenv('COHERE_API_KEY')))
print('To get one: sign in at dashboard.cohere.com, open API Keys, copy a free trial key, then put COHERE_API_KEY=... in a .env file. Never paste a key into a notebook cell or a chat.')
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm, label = get_llm()
print('model used:', label)
chain = ChatPromptTemplate.from_template('In one sentence, what is {topic}?') | llm | StrOutputParser()
print(chain.invoke({'topic': 'a prescription'}))
```

## Observations

No Cohere key was available when this was run, so the same chain ran on Groq. Only `get_llm()` changes when a key is set, which is the point of the common model interface.
