# Program 3: Five semantically similar words for any input

**Aim.** Write a program that generates 5 semantically similar words for a given input using embeddings.

**Theory.** Nearest neighbours in embedding space by cosine similarity are the words used in the most similar contexts. Similar here does not always mean synonym: antonyms (hot, cold) and related words (doctor, hospital) are also close.

```python
import sys
sys.path.insert(0, '.')
from common import *
```

```python
import gensim.downloader as api
wv = api.load('glove-wiki-gigaword-100')

def similar_words(word, n=5):
    """n nearest words, or a clear message when the word is not in the vocabulary."""
    w = word.strip().lower()
    if w not in wv:
        return None
    return [(x, round(float(s), 3)) for x, s in wv.most_similar(w, topn=n)]

# interactive use: word = input('Word: ')
for word in ['medicine', 'fever', 'hospital', 'cough', 'ibuprofen', 'qzxv']:
    r = similar_words(word)
    print(f'{word:10s}', ('not in vocabulary' if r is None else ', '.join(f'{w} ({s})' for w, s in r)))
```

```python
# similar is not the same as synonym: check an antonym pair
print('hot ~ cold', round(float(wv.similarity('hot', 'cold')), 3), '  hot ~ warm', round(float(wv.similarity('hot', 'warm')), 3))
```

## Observations

The function returns five neighbours with their cosine scores and reports unknown words instead of failing. Neighbours of a drug name are often other drugs, which is useful for suggestions but not a guarantee of equivalence, so MedSaarthi never substitutes one medicine for another.
