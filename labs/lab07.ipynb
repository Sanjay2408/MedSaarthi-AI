# Program 7: Creative sentences from similar words

**Aim.** Use embeddings to retrieve similar words for a seed word and build meaningful sentences.

**Theory.** Similar words are slotted into sentence templates so the output stays grammatical, and an LLM is asked for a freer version for comparison.

```python
import sys
sys.path.insert(0, '.')
from common import *
```

```python
import random
import gensim.downloader as api
wv = api.load('glove-wiki-gigaword-100')
llm, label = get_llm(0.7)
print('model:', label)
seed = 'healing'
words = [w for w, _ in wv.most_similar(seed, topn=12) if w.isalpha()][:5]
print('seed:', seed, '-> similar:', words)
```

```python
# the similar words can be nouns, verbs or adjectives, so each template quotes them as words
templates = ['The word "{w}" always reminds me of {seed}.', 'On the ward wall, someone had painted "{w}" next to "{seed}".',
             'She wrote "{w}" on the card that said "{seed}".', 'When we talk about {seed}, the idea of "{w}" is never far away.',
             'His diary joined "{seed}" and "{w}" in a single line.']
for w, template in zip(words, random.Random(4).sample(templates, len(words))):
    print('-', template.format(w=w, seed=seed))
```

```python
print('LLM:', ask(llm, f'Write one warm, meaningful sentence that uses all of these words: {seed}, ' + ', '.join(words) + '.'))
```

## Observations

Templates guarantee a sentence but sound repetitive; the LLM sentence is richer. Neither is checked for meaning by the program, only by reading.
